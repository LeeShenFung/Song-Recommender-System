```python
import pandas as pd

music_df = pd.read_csv("Music Info.csv")
listen_df = pd.read_csv("User Listening History.csv")
```

```python
top_users = (
    listen_df["user_id"]     # select the top 1000 most active users based on number of listening records
    .value_counts()
    .head(1000)
    .index
)

filtered_df = listen_df[
    listen_df["user_id"].isin(top_users) # keep only listening records belonging to the selected top 1000 users
]
```

```python
filtered_df.shape # show the size of the filtered dataset
```

```python
filtered_df["user_id"].nunique() # verify that the filtered dataset contains 1000 unique users
```

```python
merged_df = pd.merge(    # merge listening history with music information using track_id
    filtered_df,
    music_df,
    on="track_id",
    how="inner"
)
```

```python
merged_df.shape # check the size of the merged dataset
```

```python
user_item_matrix = merged_df.pivot_table(
    index="user_id",
    columns="track_id",
    values="playcount",
    fill_value=0
)

user_item_matrix.shape
```

```python
user_item_matrix.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(user_item_matrix)

similarity_df = pd.DataFrame(
    similarity,
    index=user_item_matrix.index,
    columns=user_item_matrix.index
)
```

```python
similarity_df.head()
```

```python
target_user = user_item_matrix.index[0] # select the first user as the target user

similar_users = (
    similarity_df[target_user]
    .drop(target_user)  #exclude target user itself.
    .sort_values(ascending=False)
    .head(5)     #top 5 user chosen
)

similar_users
```

```python
top_k_users = user_item_matrix.loc[similar_users.index] # get the listening history of the top 5 similar users
```

```python
song_scores = top_k_users.sum(axis=0)  # sum the playcounts of each track across the top 5 similar users
song_scores
```

```python
# identify tracks that the target user has not listened to

target_history = user_item_matrix.loc[target_user]
unseen_songs = target_history[target_history == 0].index
```

```python
#keep unseen songs

recommend_tracks = (
    song_scores[song_scores.index.isin(unseen_songs)]
    .sort_values(ascending=False)
    .head(10)  #top 10 recommendations based on Top 5 similar users
)

recommend_tracks 
```

```python
recommendation_result = (
    merged_df[
        merged_df["track_id"].isin(recommend_tracks.index)
    ][["track_id", "name", "artist", "genre"]]
    .drop_duplicates(subset="track_id")
    .set_index("track_id")
    .loc[recommend_tracks.index]
    .reset_index()
)

recommendation_result
```

```python
# create training and testing data for hold-out evaluation

train_matrix = user_item_matrix.copy()
test_items = {}
```

```python
# hide two previously listened tracks from each user for testing

import numpy as np

np.random.seed(42)

for user in train_matrix.index:
    listened_tracks = train_matrix.columns[
        train_matrix.loc[user] > 0
    ]

    if len(listened_tracks) > 2:
        hidden_tracks = np.random.choice(
            listened_tracks,
            size=2,
            replace=False
        )

        test_items[user] = set(hidden_tracks)

        train_matrix.loc[user, hidden_tracks] = 0
```

```python
train_similarity = cosine_similarity(train_matrix)

train_similarity_df = pd.DataFrame(
    train_similarity,
    index=train_matrix.index,
    columns=train_matrix.index
)
```

```python
def recommend_for_user(user, top_k=5, top_n=10):

    similar_users = (
        train_similarity_df[user]
        .drop(user)
        .sort_values(ascending=False)
        .head(top_k)
    )

    top_k_users = train_matrix.loc[similar_users.index]

    # weight each similar user's playcounts by their similarity score
    # weighted_scores = top_k_users.mul(
    #    similar_users,
    #    axis=0
    #)

    # combine the weighted scores for every track
    # song_scores = weighted_scores.sum(axis=0)

    song_scores = top_k_users.sum(axis=0)

    unseen_songs = train_matrix.columns[
        train_matrix.loc[user] == 0
    ]

    recommendations = (
        song_scores[
            song_scores.index.isin(unseen_songs) &
            (song_scores > 0)
        ]
        .sort_values(ascending=False)
        .head(top_n)
    )

    return recommendations.index.tolist()  # return the top 10 recommended track IDs
```

```python
# prepare test users
test_users = list(test_items.keys())

np.random.seed(42)

if len(test_users) > 300:
    test_users = np.random.choice(
        test_users,
        size=300,
        replace=False
    )

# test ONE user
test_user = test_users[0]

actual_tracks = test_items[test_user]
recommended_tracks = recommend_for_user(test_user)

print("Actual hidden tracks:")
print(actual_tracks)

print("\nRecommended tracks:")
print(recommended_tracks)

print("\nHits:")
print(actual_tracks & set(recommended_tracks))
```

```python
precision_scores = []
recall_scores = []

total_hits = 0

for user in test_users:
    relevant_tracks = test_items[user]

    recommended_tracks = set(
        recommend_for_user(
            user,
            top_k=5,
            top_n=10
        )
    )

    hits = len(relevant_tracks & recommended_tracks)

    total_hits += hits
    precision_scores.append(hits / 10)
    recall_scores.append(hits / len(relevant_tracks))

print("Total Hits:", total_hits)
print("Total Test Users:", len(test_users))
```

```python
precision_at_10 = np.mean(precision_scores)
recall_at_10 = np.mean(recall_scores)

f1_at_10 = (
    2 * precision_at_10 * recall_at_10
    / (precision_at_10 + recall_at_10)
    if (precision_at_10 + recall_at_10) > 0
    else 0
)

print("Precision@10:", precision_at_10)
print("Recall@10   :", recall_at_10)
print("F1@10       :", f1_at_10)
print("Users evaluated:", len(test_users))
```
